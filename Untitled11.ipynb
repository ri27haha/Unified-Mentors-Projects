```python
# 1. Install libraries and load dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load standard enterprise HR workforce attrition dataset
url = "https://raw.githubusercontent.com/taknev83/datasets/master/HR_Employee_Attrition.csv"
df = pd.read_csv(url)

print("Dataset Shape:", df.shape)
print("\nFirst 3 rows:")
display(df.head(3))

# 2. Data Validation & Cleaning (Map Attrition to Binary/Numeric for analysis)
df['Attrition_Num'] = df['Attrition'].apply(lambda x: 1 if str(x).strip().lower() == 'yes' else 0)

# 3. Overall Attrition Assessment (KPIs)
total_employees = len(df)
total_exits = df['Attrition_Num'].sum()
overall_attrition_rate = (total_exits / total_employees) * 100

print("\n--- EXECUTIVE WORKFORCE KPIs ---")
print(f"Total Employees Tracked: {total_employees}")
print(f"Total Exits (Attrition): {total_exits}")
print(f"Overall Attrition Rate: {overall_attrition_rate:.2f}%")

# 4. Visualizations for Research Paper & Reports
sns.set_theme(style="whitegrid")

# Fig 1: Overall Attrition Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Attrition', palette='Set2')
plt.title('Overall Attrition vs Retention Count')
plt.show()

# Fig 2: Department-wise Attrition Rate Hotspots
plt.figure(figsize=(10, 5))
dept_attrition = df.groupby('Department')['Attrition_Num'].mean().reset_index()
dept_attrition['Attrition_Rate_%'] = dept_attrition['Attrition_Num'] * 100
sns.barplot(data=dept_attrition, x='Department', y='Attrition_Rate_%', palette='rocket')
plt.title('Attrition Rate Hotspots by Department (%)')
plt.ylabel('Attrition Rate (%)')
plt.show()

# Fig 3: Job Role Attrition Frequency
plt.figure(figsize=(12, 5))
role_attrition = df.groupby('JobRole')['Attrition_Num'].mean().reset_index()
role_attrition['Attrition_Rate_%'] = role_attrition['Attrition_Rate_%'] = role_attrition['Attrition_Num'] * 100
role_attrition = role_attrition.sort_values(by='Attrition_Rate_%', ascending=False)
sns.barplot(data=role_attrition, x='Attrition_Rate_%', y='JobRole', palette='mako')
plt.title('Attrition Rate by Specific Job Role (%)')
plt.xlabel('Attrition Rate (%)')
plt.show()

# Fig 4: Tenure Buckets (Years At Company) vs Attrition
bins = [-1, 2, 5, 10, 40]
labels = ['Early Career (0-2 yrs)', 'Mid-Early (3-5 yrs)', 'Established (6-10 yrs)', 'Veteran (10+ yrs)']
df['TenureBucket'] = pd.cut(df['YearsAtCompany'], bins=bins, labels=labels)

plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='TenureBucket', y='Attrition_Num', ci=None, palette='viridis')
plt.title('Attrition Likelihood by Company Tenure Stage')
plt.ylabel('Attrition Probability Ratio')
plt.show()

# Fig 5: Workload Impact (OverTime vs Attrition)
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='OverTime', y='Attrition_Num', palette='coolwarm', ci=None)
plt.title('Impact of OverTime on Attrition Rate')
plt.ylabel('Attrition Probability')
plt.show()
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import plotly.express as px

st.set_page_config(page_title="Palo Alto Networks - Workforce Attrition Analytics", layout="wide")
st.title("🛡️ Palo Alto Networks: Workforce Attrition & Risk Hotspot Dashboard")
st.markdown("Diagnostic HR intelligence platform tracking employee turnover drivers, department risk levels, and workload impacts.")

@st.cache_data
def load_data():
    url = "https://raw.githubusercontent.com/taknev83/datasets/master/HR_Employee_Attrition.csv"
    df = pd.read_csv(url)
    df['Attrition_Num'] = df['Attrition'].apply(lambda x: 1 if str(x).strip().lower() == 'yes' else 0)
    return df

df = load_data()

# Sidebar Filters
st.sidebar.header("Workforce Filter Panel")
selected_dept = st.sidebar.multiselect("Select Department", df['Department'].unique(), default=df['Department'].unique())
selected_role = st.sidebar.multiselect("Select Job Role", df['JobRole'].unique(), default=df['JobRole'].unique())
selected_travel = st.sidebar.multiselect("Business Travel Frequency", df['BusinessTravel'].unique(), default=df['BusinessTravel'].unique())
selected_ot = st.sidebar.multiselect("OverTime Status", df['OverTime'].unique(), default=df['OverTime'].unique())

max_tenure = int(df['YearsAtCompany'].max())
tenure_range = st.sidebar.slider("Years at Company Range", 0, max_tenure, (0, max_tenure))

# Apply Filters
filtered_df = df[
    (df['Department'].isin(selected_dept)) &
    (df['JobRole'].isin(selected_role)) &
    (df['BusinessTravel'].isin(selected_travel)) &
    (df['OverTime'].isin(selected_ot)) &
    (df['YearsAtCompany'] >= tenure_range[0]) &
    (df['YearsAtCompany'] <= tenure_range[1])
]

# KPI Metrics Bar
st.subheader("Key Workforce Performance Indicators")
col1, col2, col3, col4 = st.columns(4)
total_emp = len(filtered_df)
total_attrition = filtered_df['Attrition_Num'].sum()
att_rate = (total_attrition / total_emp * 100) if total_emp > 0 else 0

col1.metric("Total Filtered Workforce", total_emp)
col2.metric("Total Exits", total_attrition)
col3.metric("Attrition Rate (%)", f"{att_rate:.2f}%")
col4.metric("Avg Monthly Income", f"${filtered_df['MonthlyIncome'].mean():,.0f}" if total_emp > 0 else "$0")

st.divider()

# Layout Charts
col_a, col_b = st.columns(2)

with col_a:
    st.subheader("Department Risk Hotspots")
    if not filtered_df.empty:
        dept_grp = filtered_df.groupby('Department')['Attrition_Num'].mean().reset_index()
        dept_grp['Attrition Rate'] = dept_grp['Attrition_Num'] * 100
        fig_dept = px.bar(dept_grp, x='Department', y='Attrition Rate', color='Attrition Rate', title="Attrition % by Department")
        st.plotly_chart(fig_dept, use_container_width=True)
    else:
        st.warning("No data available for current filter selection.")

with col_b:
    st.subheader("OverTime vs Attrition Impact")
    if not filtered_df.empty:
        ot_grp = filtered_df.groupby('OverTime')['Attrition_Num'].mean().reset_index()
        ot_grp['Attrition Rate'] = ot_grp['Attrition_Num'] * 100
        fig_ot = px.bar(ot_grp, x='OverTime', y='Attrition Rate', color='OverTime', title="Attrition Probability: OverTime vs Non-OverTime")
        st.plotly_chart(fig_ot, use_container_width=True)
    else:
        st.warning("No data available for current filter selection.")

col_c, col_d = st.columns(2)

with col_c:
    st.subheader("Tenure Distribution of Exits")
    if not filtered_df.empty:
        exits_df = filtered_df[filtered_df['Attrition_Num'] == 1]
        fig_tenure = px.histogram(exits_df, x='YearsAtCompany', nbins=10, title="Exits Grouped by Years at Company")
        st.plotly_chart(fig_tenure, use_container_width=True)
    else:
        st.warning("No exits found in selection.")

with col_d:
    st.subheader("Monthly Income vs Attrition")
    if not filtered_df.empty:
        fig_income = px.box(filtered_df, x='Attrition', y='MonthlyIncome', color='Attrition', title="Income Distribution by Attrition Status")
        st.plotly_chart(fig_income, use_container_width=True)
    else:
        st.warning("No data available for current filter selection.")
```
