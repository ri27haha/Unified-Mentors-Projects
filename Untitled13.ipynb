```python
# Function to download and load dataset from Google Drive
def load_data():
    file_id = "19ThyYpOoBtBvujzJvvvw3wgEjhjLRkQ0"
    url = f"https://drive.google.com/uc?id={file_id}"
    output = "employee_data.csv"

    if not os.path.exists(output):
        print("Downloading dataset...")
        try:
            !gdown --id {file_id} -O {output}
        except Exception as e:
            print("gdown failed, trying standard download...", e)
            urllib.request.urlretrieve(url, output)

    df = pd.read_csv(output)
    return df

df = load_data()
print(f"Dataset loaded successfully with shape: {df.shape}")

# --- Feature Engineering ---
# Avoid division by zero by adding small epsilon or handling zero tenure
df['YearsAtCompany_Safe'] = df['YearsAtCompany'].replace(0, 0.5)

df['Promotion_Gap_Ratio'] = df['YearsSinceLastPromotion'] / df['YearsAtCompany_Safe']
df['Role_Stagnation_Index'] = df['YearsInCurrentRole'] / df['YearsAtCompany_Safe']
df['Training_Intensity_Score'] = df['TrainingTimesLastYear'] / df['YearsAtCompany_Safe']
df['Manager_Stability_Ratio'] = df['YearsWithCurrManager'] / df['YearsAtCompany_Safe']

# --- Promotion Gap Risk Scoring ---
def assign_promotion_gap_risk(row):
    gap = row['YearsSinceLastPromotion']
    stagnation = row['Role_Stagnation_Index']

    if gap >= 4 or stagnation >= 0.8:
        return 'High'
    elif gap >= 2 or stagnation >= 0.5:
        return 'Medium'
    else:
        return 'Low'

df['Promotion_Gap_Score'] = df.apply(assign_promotion_gap_risk, axis=1)

# --- Unsupervised Clustering (K-Means) ---
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

cluster_features = [
    'Promotion_Gap_Ratio',
    'Role_Stagnation_Index',
    'Training_Intensity_Score',
    'YearsAtCompany',
    'JobLevel',
    'PercentSalaryHike'
]

X = df[cluster_features].fillna(0)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit K-Means with 4 clusters representing career archetypes
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['Cluster_ID'] = kmeans.fit_predict(X_scaled)

# Map clusters to descriptive business labels based on centroids
cluster_mapping = {
    0: 'Stable Contributors',
    1: 'Promotion-Stalled / High Risk',
    2: 'Fast-Track / Early Explorers',
    3: 'Tenured Specialists'
}
df['Career_Cluster'] = df['Cluster_ID'].map(cluster_mapping)

# --- Retention Opportunity Identification ---
# Flag employees who show stagnation signals but haven't actively disengaged (e.g., JobSatisfaction >= 3, Attrition == 0)
def identify_retention_opportunity(row):
    is_stagnant = 1 if row['Promotion_Gap_Score'] in ['Medium', 'High'] else 0
    is_satisfied = 1 if row.get('JobSatisfaction', 3) >= 3 else 0
    not_left = 1 if row.get('Attrition', 0) == 0 else 0

    if is_stagnant == 1 and is_satisfied == 1 and not_left == 1:
        return 'High Opportunity'
    elif is_stagnant == 1:
        return 'Monitor'
    else:
        return 'Stable'

df['Retention_Opportunity_Status'] = df.apply(identify_retention_opportunity, axis=1)

print("Career features, clustering, and risk scoring completed successfully!")
display(df[['Age', 'Department', 'JobRole', 'YearsSinceLastPromotion', 'Career_Cluster', 'Promotion_Gap_Score', 'Retention_Opportunity_Status']].head())
```

```python
app_code = """
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

st.set_page_config(
    page_title="Palo Alto Networks - Career Progression & Retention",
    page_icon="🚀",
    layout="wide"
)

@st.cache_data
def load_and_process_data():
    try:
        df = pd.read_csv("employee_data.csv")
    except:
        return None

    df['YearsAtCompany_Safe'] = df['YearsAtCompany'].replace(0, 0.5)
    df['Promotion_Gap_Ratio'] = df['YearsSinceLastPromotion'] / df['YearsAtCompany_Safe']
    df['Role_Stagnation_Index'] = df['YearsInCurrentRole'] / df['YearsAtCompany_Safe']
    df['Training_Intensity_Score'] = df['TrainingTimesLastYear'] / df['YearsAtCompany_Safe']

    def assign_promotion_gap_risk(row):
        gap = row['YearsSinceLastPromotion']
        stagnation = row['Role_Stagnation_Index']
        if gap >= 4 or stagnation >= 0.8:
            return 'High'
        elif gap >= 2 or stagnation >= 0.5:
            return 'Medium'
        else:
            return 'Low'

    df['Promotion_Gap_Score'] = df.apply(assign_promotion_gap_risk, axis=1)

    from sklearn.preprocessing import StandardScaler
    from sklearn.cluster import KMeans
    cluster_features = ['Promotion_Gap_Ratio', 'Role_Stagnation_Index', 'Training_Intensity_Score', 'YearsAtCompany', 'JobLevel', 'PercentSalaryHike']
    X = df[cluster_features].fillna(0)
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
    df['Cluster_ID'] = kmeans.fit_predict(X_scaled)

    cluster_mapping = {
        0: 'Stable Contributors',
        1: 'Promotion-Stalled / High Risk',
        2: 'Fast-Track / Early Explorers',
        3: 'Tenured Specialists'
    }
    df['Career_Cluster'] = df['Cluster_ID'].map(cluster_mapping)

    def identify_retention_opportunity(row):
        is_stagnant = 1 if row['Promotion_Gap_Score'] in ['Medium', 'High'] else 0
        is_satisfied = 1 if row.get('JobSatisfaction', 3) >= 3 else 0
        not_left = 1 if row.get('Attrition', 0) == 0 else 0
        if is_stagnant == 1 and is_satisfied == 1 and not_left == 1:
            return 'High Opportunity'
        elif is_stagnant == 1:
            return 'Monitor'
        else:
            return 'Stable'

    df['Retention_Opportunity_Status'] = df.apply(identify_retention_opportunity, axis=1)
    return df

df = load_and_process_data()

if df is None:
    st.error("Dataset 'employee_data.csv' not found. Please run the data preparation cell first!")
else:
    # --- SIDEBAR CONTROLS ---
    st.sidebar.header("🚀 Career Intelligence Filters")

    departments = ['All'] + list(df['Department'].unique())
    selected_dept = st.sidebar.selectbox("Select Department", departments)

    job_levels = ['All'] + sorted(list(df['JobLevel'].unique()))
    selected_level = st.sidebar.selectbox("Select Job Level", job_levels)

    promo_gap_filter = st.sidebar.selectbox("Promotion Gap Score Filter", ['All', 'Low', 'Medium', 'High'])

    max_years = int(df['YearsSinceLastPromotion'].max()) if 'YearsSinceLastPromotion' in df.columns else 10
    promo_threshold = st.sidebar.slider("Min Years Since Last Promotion", 0, max_years, 0)

    # Apply filters
    filtered_df = df.copy()
    if selected_dept != 'All':
        filtered_df = filtered_df[filtered_df['Department'] == selected_dept]
    if selected_level != 'All':
        filtered_df = filtered_df[filtered_df['JobLevel'] == selected_level]
    if promo_gap_filter != 'All':
        filtered_df = filtered_df[filtered_df['Promotion_Gap_Score'] == promo_gap_filter]
    filtered_df = filtered_df[filtered_df['YearsSinceLastPromotion'] >= promo_threshold]

    # --- HEADER ---
    st.title("Palo Alto Networks - Career Progression & Promotion Gap Analysis")
    st.markdown("**Proactive Retention Intelligence**: Uncovering structural career stagnation and promotion bottlenecks before disengagement occurs.")

    # Top Metrics Row
    c1, c2, c3, c4 = st.columns(4)
    with c1:
        st.metric("Analyzed Workforce", len(filtered_df))
    with c2:
        stagnant_count = len(filtered_df[filtered_df['Promotion_Gap_Score'] == 'High'])
        st.metric("High Promotion Gap Count", stagnant_count, delta_color="inverse")
    with c3:
        high_opp_count = len(filtered_df[filtered_df['Retention_Opportunity_Status'] == 'High Opportunity'])
        st.metric("Retention Opportunities", high_opp_count)
    with c4:
        avg_promo_years = round(filtered_df['YearsSinceLastPromotion'].mean(), 1) if len(filtered_df) > 0 else 0
        st.metric("Avg Years Since Promotion", f"{avg_promo_years} yrs")

    st.markdown("---")

    # --- TABS FOR MODULES ---
    tab1, tab2, tab3, tab4 = st.tabs([
        "🧬 Career Path Clustering",
        "⏳ Promotion Gap Monitor",
        "💡 Retention Opportunity Panel",
        "👥 Managerial & Team Insights"
    ])

    with tab1:
        st.subheader("Unsupervised Career Path Clustering & Archetypes")
        c1, c2 = st.columns(2)
        with c1:
            st.markdown("**Career Cluster Distribution**")
            cluster_counts = filtered_df['Career_Cluster'].value_counts()
            fig, ax = plt.subplots(figsize=(6, 4))
            cluster_counts.plot(kind='barh', color='darkorange', ax=ax)
            ax.set_xlabel("Employee Count")
            st.pyplot(fig)
        with c2:
            st.markdown("**Tenure vs. Promotion Gap by Cluster**")
            fig, ax = plt.subplots(figsize=(6, 4))
            sns.scatterplot(x='YearsAtCompany', y='YearsSinceLastPromotion', hue='Career_Cluster', data=filtered_df, palette='Set2', ax=ax)
            st.pyplot(fig)

    with tab2:
        st.subheader("Promotion Stagnation & Role Gap Analysis")
        c1, c2 = st.columns(2)
        with c1:
            st.markdown("**Promotion Gap Risk Breakdown**")
            risk_counts = filtered_df['Promotion_Gap_Score'].value_counts()
            fig, ax = plt.subplots(figsize=(5, 5))
            ax.pie(risk_counts, labels=risk_counts.index, autopct='%1.1f%%', colors=['#2ca02c','#ff7f0e','#d62728'], startangle=90)
            st.pyplot(fig)
        with c2:
            st.markdown("**Role Stagnation Index by Job Role**")
            if 'JobRole' in filtered_df.columns:
                role_stag = filtered_df.groupby('JobRole')['Role_Stagnation_Index'].mean().sort_values()
                fig, ax = plt.subplots(figsize=(6, 4))
                role_stag.plot(kind='bar', color='purple', ax=ax)
                ax.set_ylabel("Avg Role Stagnation Index")
                plt.xticks(rotation=45, ha='right')
                st.pyplot(fig)

    with tab3:
        st.subheader("Proactive Retention Opportunity Panel")
        st.markdown("Employees exhibiting career stagnation signals who remain satisfied and un-departed—representing prime windows for proactive career interventions:")

        opp_df = filtered_df[filtered_df['Retention_Opportunity_Status'] == 'High Opportunity'][[
            'Age', 'Department', 'JobRole', 'JobLevel', 'YearsAtCompany', 'YearsSinceLastPromotion', 'TrainingTimesLastYear', 'Promotion_Gap_Score'
        ]]
        st.dataframe(opp_df, use_container_width=True)

        csv = opp_df.to_csv(index=False).encode('utf-8')
        st.download_button(
            label="Download Retention Opportunity Action List (CSV)",
            data=csv,
            file_name='palo_alto_retention_opportunities.csv',
            mime='text/csv',
        )

    with tab4:
        st.subheader("Managerial Continuity & Team Stagnation Insights")
        if 'YearsWithCurrManager' in filtered_df.columns:
            fig, ax = plt.subplots(figsize=(10, 4))
            sns.boxplot(x='JobLevel', y='YearsWithCurrManager', hue='Promotion_Gap_Score', data=filtered_df, ax=ax)
            ax.set_title("Managerial Tenure vs. Promotion Gap Across Job Levels")
            st.pyplot(fig)
"""

with open("app.py", "w") as f:
    f.write(app_code)

print("Streamlit app file 'app.py' created successfully for Career Progression Project!")
```
