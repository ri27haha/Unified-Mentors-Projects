```python
# Function to download and load dataset from Google Drive
def load_data():
    file_id = "19ThyYpOoBtBvujzJvvvw3wgEjhjLRkQ0"
    url = f"https://drive.google.com/uc?id={file_id}"
    output = "employee_data.csv"

    if not os.path.exists(output):
        print("Downloading dataset...")
        try:
            # Using gdown if available or standard url opener for Google Drive large files
            !gdown --id {file_id} -O {output}
        except Exception as e:
            print("gdown failed, trying standard download...", e)
            urllib.request.urlretrieve(url, output)

    df = pd.read_csv(output)
    return df

df = load_data()
print(f"Dataset loaded successfully with shape: {df.shape}")

# --- Data Validation & Normalization ---
# Validate and handle missing values for satisfaction metrics (scale 1-4)
satisfaction_cols = ['EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'JobInvolvement']
for col in satisfaction_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())
        # Normalize to 0-1 scale for composite index comparability
        df[f'{col}_Norm'] = (df[col] - 1) / 3.0

# --- Engagement Index Construction ---
norm_cols = [f'{col}_Norm' for col in satisfaction_cols]
df['Engagement_Index'] = df[norm_cols].mean(axis=1) * 100 # Scaled 0 to 100

# --- Burnout Risk Identification ---
# Flag employees with OverTime == 'Yes' and WorkLifeBalance <= 2
def calculate_burnout(row):
    ot = 1 if str(row.get('OverTime', '')).strip().lower() == 'yes' else 0
    wlb = row.get('WorkLifeBalance', 3)

    if ot == 1 and wlb <= 2:
        return 'High'
    elif ot == 1 or wlb <= 2:
        return 'Medium'
    else:
        return 'Low'

df['Burnout_Risk'] = df.apply(calculate_burnout, axis=1)

# Numeric Burnout Score for tracking
risk_mapping = {'Low': 1, 'Medium': 2, 'High': 3}
df['Burnout_Risk_Score'] = df['Burnout_Risk'].map(risk_mapping)

print("Feature engineering completed successfully!")
display(df[['Age', 'Department', 'JobRole', 'Engagement_Index', 'Burnout_Risk']].head())
```

```python
app_code = """
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

st.set_page_config(
    page_title="Palo Alto Networks - Employee Experience Diagnostic",
    page_icon="🛡️",
    layout="wide"
)

@st.cache_data
def load_data():
    # Load processed dataframe or raw dataset
    try:
        df = pd.read_csv("employee_data.csv")
    except:
        # Fallback dummy sample if file path differs
        return None

    # Re-apply feature engineering
    satisfaction_cols = ['EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'JobInvolvement']
    for col in satisfaction_cols:
        if col in df.columns:
            df[col] = df[col].fillna(df[col].median())
            df[f'{col}_Norm'] = (df[col] - 1) / 3.0

    norm_cols = [f'{col}_Norm' for col in satisfaction_cols if f'{col}_Norm' in df.columns]
    if norm_cols:
        df['Engagement_Index'] = df[norm_cols].mean(axis=1) * 100
    else:
        df['Engagement_Index'] = 50.0

    def calculate_burnout(row):
        ot = 1 if str(row.get('OverTime', '')).strip().lower() == 'yes' else 0
        wlb = row.get('WorkLifeBalance', 3)
        if ot == 1 and wlb <= 2:
            return 'High'
        elif ot == 1 or wlb <= 2:
            return 'Medium'
        else:
            return 'Low'

    df['Burnout_Risk'] = df.apply(calculate_burnout, axis=1)
    return df

df = load_data()

if df is None:
    st.error("Dataset 'employee_data.csv' not found. Please run the data preparation cell first!")
else:
    # --- SIDEBAR FILTERS ---
    st.sidebar.header("🛡️ Diagnostics Controls")

    departments = ['All'] + list(df['Department'].unique())
    selected_dept = st.sidebar.selectbox("Select Department", departments)

    ot_options = ['All', 'Yes', 'No']
    selected_ot = st.sidebar.selectbox("OverTime Status", ot_options)

    engagement_threshold = st.sidebar.slider("Engagement Index Threshold Max", 0, 100, 100)

    tenure_max = int(df['YearsAtCompany'].max()) if 'YearsAtCompany' in df.columns else 40
    tenure_range = st.sidebar.slider("Years at Company Range", 0, tenure_max, (0, tenure_max))

    # Apply filters
    filtered_df = df.copy()
    if selected_dept != 'All':
        filtered_df = filtered_df[filtered_df['Department'] == selected_dept]
    if selected_ot != 'All':
        filtered_df = filtered_df[filtered_df['OverTime'].str.strip().str.lower() == selected_ot.lower()]
    filtered_df = filtered_df[
        (filtered_df['Engagement_Index'] <= engagement_threshold) &
        (filtered_df['YearsAtCompany'] >= tenure_range[0]) &
        (filtered_df['YearsAtCompany'] <= tenure_range[1])
    ]

    # --- MAIN DASHBOARD HEADER ---
    st.title("Palo Alto Networks - Employee Engagement & Burnout Diagnostics")
    st.markdown("**Proactive HR Analytics Dashboard**: Shifting from reactive attrition analysis to early-warning well-being interventions.")

    # Top KPI Metrics Row
    col1, col2, col3, col4 = st.columns(4)
    with col1:
        st.metric("Total Analyzed Employees", len(filtered_df))
    with col2:
        avg_eng = round(filtered_df['Engagement_Index'].mean(), 1) if len(filtered_df) > 0 else 0
        st.metric("Avg Engagement Index", f"{avg_eng}/100")
    with col3:
        high_burnout_count = len(filtered_df[filtered_df['Burnout_Risk'] == 'High'])
        st.metric("High Burnout Risk Count", high_burnout_count, delta_color="inverse")
    with col4:
        attrition_rate = round((filtered_df['Attrition'].sum() / len(filtered_df)) * 100, 1) if 'Attrition' in filtered_df.columns and len(filtered_df) > 0 else 0
        st.metric("Attrition Rate", f"{attrition_rate}%")

    st.markdown("---")

    # --- TABS FOR MODULES ---
    tab1, tab2, tab3, tab4 = st.tabs([
        "📊 Engagement Health Overview",
        "🔥 Burnout Risk Dashboard",
        "👥 Role & Career Stage Analysis",
        "🚨 Manager Action Panel"
    ])

    with tab1:
        st.subheader("Organization-Wide Engagement Health")
        c1, c2 = st.columns(2)
        with c1:
            st.markdown("**Engagement Index Distribution**")
            fig, ax = plt.subplots(figsize=(6, 4))
            sns.histplot(filtered_df['Engagement_Index'], kde=True, color='teal', ax=ax)
            ax.set_xlabel("Engagement Index Score")
            ax.set_ylabel("Employee Count")
            st.pyplot(fig)
        with c2:
            st.markdown("**Satisfaction Dimensions Breakdown**")
            sat_means = filtered_df[['EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'JobInvolvement']].mean()
            fig, ax = plt.subplots(figsize=(6, 4))
            sat_means.plot(kind='barh', color='skyblue', ax=ax)
            ax.set_xlim(1, 4)
            ax.set_xlabel("Average Rating (1-4 Scale)")
            st.pyplot(fig)

    with tab2:
        st.subheader("Burnout Risk Breakdown & Imbalance Factors")
        c1, c2 = st.columns(2)
        with c1:
            st.markdown("**Burnout Risk Level Proportions**")
            risk_counts = filtered_df['Burnout_Risk'].value_counts()
            fig, ax = plt.subplots(figsize=(5, 5))
            ax.pie(risk_counts, labels=risk_counts.index, autopct='%1.1f%%', colors=['#66b3ff','#99ff99','#ff9999'], startangle=90)
            st.pyplot(fig)
        with c2:
            st.markdown("**Overtime vs Work-Life Balance Impact**")
            if 'WorkLifeBalance' in filtered_df.columns and 'OverTime' in filtered_df.columns:
                fig, ax = plt.subplots(figsize=(6, 4))
                sns.boxplot(x='OverTime', y='Engagement_Index', hue='WorkLifeBalance', data=filtered_df, ax=ax)
                ax.set_title("Engagement across Overtime & Work-Life Balance")
                st.pyplot(fig)

    with tab3:
        st.subheader("Career Stage & Job Role Disengagement Analysis")
        if 'JobRole' in filtered_df.columns:
            role_eng = filtered_df.groupby('JobRole')['Engagement_Index'].mean().sort_values()
            fig, ax = plt.subplots(figsize=(10, 5))
            role_eng.plot(kind='bar', color='coral', ax=ax)
            ax.set_ylabel("Average Engagement Index")
            ax.set_title("Engagement Index by Job Role")
            plt.xticks(rotation=45, ha='right')
            st.pyplot(fig)

        st.markdown("**Tenure vs. Engagement Trends**")
        if 'YearsAtCompany' in filtered_df.columns:
            fig, ax = plt.subplots(figsize=(10, 4))
            sns.lineplot(x='YearsAtCompany', y='Engagement_Index', data=filtered_df, ci=None, ax=ax, color='purple')
            ax.set_title("Engagement Trend over Years at Palo Alto Networks")
            st.pyplot(fig)

    with tab4:
        st.subheader("Manager Action Panel - Priority Interventions")
        st.markdown("Employees flagged below with **High Burnout Risk** or **Critical Low Engagement (< 40)** requiring immediate manager intervention:")

        intervention_df = filtered_df[
            (filtered_df['Burnout_Risk'] == 'High') |
            (filtered_df['Engagement_Index'] < 40)
        ][['Age', 'Department', 'JobRole', 'JobLevel', 'MonthlyIncome', 'OverTime', 'WorkLifeBalance', 'Engagement_Index', 'Burnout_Risk']]

        st.dataframe(intervention_df, use_container_width=True)

        csv = intervention_df.to_csv(index=False).encode('utf-8')
        st.download_button(
            label="Download Priority Intervention Action List (CSV)",
            data=csv,
            file_name='palo_alto_priority_interventions.csv',
            mime='text/csv',
        )
"""

with open("app.py", "w") as f:
    f.write(app_code)

print("Streamlit app file 'app.py' created successfully!")
```
